```python
from chonkie import RecursiveChunker
import chromadb
import glob, os, uuid

chunker = RecursiveChunker.from_recipe("markdown")
```

```python
chroma_client = chromadb.PersistentClient(path="./chroma_langchain_db")

collection = chroma_client.get_collection(name="documentation")
```

```python

# Replace this with the path to your main folder
main_folder = 'files'

# Loop through all subfolders and files
def create_collection_adds(path: str) -> dict:
    collection_adds = {
                'ids': [],
                'documents': [],
                'metadatas': [],
            }
    for root, dirs, files in os.walk(main_folder):
        for file in files:
            if file.endswith('.md'):
                file_path = os.path.join(root, file)
                with open(file_path, 'r', encoding='utf-8') as f:
                    content = f.read()
                    chunks = chunker(content)
                    for i, chunk in enumerate(chunks):
                        collection_adds['ids'].append(str(uuid.uuid4()) + "-"+ root.replace(f'{main_folder}/', ""))
                        collection_adds['documents'].append(chunk.text)
                        collection_adds['metadatas'].append({ "product": root.replace(f'{main_folder}/', "")})
    return collection_adds

```

```python
collection_adds = create_collection_adds(main_folder)
```
